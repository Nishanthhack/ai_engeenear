```python
stu = [['nis', 80],
       ['bhav',95],
       ['keer',50],
       ['bhar',70]]

print(list(filter(lambda row : row[0][0] == 'b',stu)))
```

```python
stu = [
    ['nis', 80],
    ['ram', 95],
    ['keer', 50],
    ['bhar', 70]
]

result = sorted(stu, key=lambda row: row[1])

print(result)
```

```python
print(list(sorted(stu , key = lambda row : len(row[0]) )))
```

```python
nums = [12, 5, 18, 7, 24, 9, 30, 11]

# Your code here
# Keep numbers greater than 10 and multiply them by 2
result = [
    d*2
    for d in nums
    if d >  10
]
print(result)
```

```python
students = [
    ["Nis", 85],
    ["Bhav", 45],
    ["Keer", 92],
    ["Bhar", 60],
    ["Arun", 35]
]

# Your code here
# Extract names of students who scored 60 or above
result = [
    row[0]
    for row in students
    if row[1] >= 60
]
print(result)
```

```python
words = ["python", "AI", "developer", "code", "machine", "learning"]

# Your code here
# Get lengths of words containing more than 4 characters
result = [
    len(d)
    for d in words
    if len(d) > 4
]

print(result)
```

```python
nums = [2, 3, 2, 4, 3, 5, 4, 6]

# Your code here
# Keep unique numbers in their original order
# Then square each number
result = [
    d**2
    for i,d in enumerate(nums)
    if d not in nums[:i]
]

print(result)
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Your code here
# Extract only even numbers from the matrix
# Square each extracted number
result = [
    d**2
    for row in matrix
    for d in row
    if d % 2 == 0
]

print(result)
```
