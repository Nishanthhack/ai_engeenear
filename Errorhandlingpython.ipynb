```python
values = ["10", "20", "abc", "40"]

# Convert each value to an integer
# If conversion fails, print the invalid value
# Continue processing the remaining values
for value in values:
  try:
    print(int(value))
  except ValueError :
    print("Invalid value:",value)





print("Processing completed")
```

```python
nums = [10, 20, 30, 40]
index = 6

# Try accessing nums[index]
# Handle the error and print "Index out of range"
try:
  print(nums[5])
except IndexError:
  print("Index out of range")

print("Done")
```

```python
a = "25"
b = 'g'

# Convert a and b to integers
# Divide a by b
# Handle ValueError and ZeroDivisionError separately
# Print a suitable message for each error
try:
  c = int(a)/int(b)
except ValueError:
  print("Invalid value")
except ZeroDivisionError:
  print("zero division error")

print("End of program")
```
